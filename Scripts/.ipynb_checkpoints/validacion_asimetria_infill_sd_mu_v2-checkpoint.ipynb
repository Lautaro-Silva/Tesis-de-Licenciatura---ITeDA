```python
# =============================================================================
# CELDA 1: Importaciones, Estilo Visual y Funciones Auxiliares
# =============================================================================
import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings

warnings.simplefilter('ignore')

# Configuración visual global (Estilo HEP)
hep_style = {
    'font.family': 'sans-serif',
    'font.sans-serif': ['Helvetica', 'Arial', 'DejaVu Sans'],
    'xtick.direction': 'in', 'ytick.direction': 'in',
    'xtick.top': True, 'ytick.right': True,
    'xtick.minor.visible': True, 'ytick.minor.visible': True,
    'axes.linewidth': 1.2, 'legend.frameon': True,
    'font.size': 14, 'axes.labelsize': 14, 'axes.titlesize': 16, 
    'figure.figsize': (8, 6)
}
plt.rcParams.update(hep_style)

# Funciones Auxiliares
def parse_metadata(folder_name):
    """Extrae metadatos basados en el nombre de la carpeta."""
    f = folder_name.lower()
    model = 'SIBYLL' if 'sib' in f else 'QGSJet' if 'qgs' in f else 'EPOS' if 'epos' in f else 'Unknown'
    if 'proton' in f: primary = 'Proton'
    elif 'hierro' in f or 'iron' in f: primary = 'Hierro'
    elif 'helio' in f or 'helium' in f: primary = 'Helio'
    elif 'oxigeno' in f or 'oxygen' in f: primary = 'Oxigeno'
    else: primary = 'Unknown'
    energy = '18.0-18.5' if '_18' in f else '17.5-18.0'
    return model, primary, energy

def ensure_degrees(series, name="Var"):
    """Convierte a grados si los valores máximos indican que está en radianes."""
    if series.dropna().abs().max() < 7.0: 
        return np.rad2deg(series)
    return series
```

```python
# =============================================================================
# CELDA 2: Carga de Datos y Creación del Dataset Guía
# =============================================================================
ROOT_DIR = "/home/lsilva/Github/ADST_Alexey_module_v9" 

all_dfs = []
if os.path.exists(ROOT_DIR):
    for folder in os.listdir(ROOT_DIR):
        folder_path = os.path.join(ROOT_DIR, folder)
        if not os.path.isdir(folder_path): continue
            
        model, primary, energy = parse_metadata(folder)
        parquets = glob.glob(os.path.join(folder_path, "*.parquet"))
        
        for f in parquets:
            try:
                # Levantamos todas las columnas necesarias, incluyendo las componentes del SD
                df = pd.read_parquet(f, columns=[
                    'event_id', 'counterId', 'theta_REC', 'theta_MC', 'r_core',
                    'nMuones_REC', 'nMuones_MC', 'sdSignal_REC', 'sd_nMuons_MC', 'sd_nEM_MC', 
                    'phi_plane_sp', 'phi_plane_euler', 'phi_plane_ground', 'module_status'
                ])
                # Añadir metadatos
                df['Model'], df['Primary'], df['Energy'] = model, primary, energy
                all_dfs.append(df)
            except Exception as e: 
                pass

    df_master = pd.concat(all_dfs, ignore_index=True)
else:
    print(f"Ruta {ROOT_DIR} no encontrada. Modificá ROOT_DIR si estás en otra PC.")
    df_master = pd.DataFrame() # Fallback

# Filtramos columnas duplicadas para evitar que Pandas explote (ValueError)
df_master = df_master.loc[:, ~df_master.columns.duplicated()].copy()

# Aislamos el Dataset Guía para este análisis (SIBYLL Proton 17.5-18.0)
df_guia = df_master[(df_master['Model'] == 'SIBYLL') & 
                    (df_master['Primary'] == 'Proton') & 
                    (df_master['Energy'] == '17.5-18.0')].copy()

df_guia = df_guia.reset_index(drop=True)
print(f"Dataset Guía cargado y limpio. Total de módulos: {len(df_guia):,}")
```

```python
# =========================================================================
# CELDA: PERFILES INFILL (40°-60°) - 5 FILAS (UMD + SD DECOUPLED)
# =========================================================================

# 1. Filtrar Infill y Rango Zenital Inclinado
df_inf = df_guia[(df_guia['counterId'] >= 100000) & 
                 (df_guia['theta_REC'] >= 40) & 
                 (df_guia['theta_REC'] <= 60)].copy()

# 2. Configuración de Bines Radiales (6 columnas)
r_edges = [0, 200, 400, 600, 800, 1000, 1500] 
r_labels = ["0-200 m", "200-400 m", "400-600 m", "600-800 m", "800-1000 m", "1000-1500 m"]
df_inf['r_bin'] = pd.cut(df_inf['r_core'], bins=r_edges, labels=False, include_lowest=True)

# 3. Pre-cálculo de ángulos con Euler Fix (REC y MC)
p_rec = ensure_degrees(df_inf['phi_plane_euler'], 'Euler REC')
df_inf['p_rec_cen'] = (p_rec + 180) % 360 - 180

if 'phi_plane_euler_MC' in df_inf.columns:
    p_mc = ensure_degrees(df_inf['phi_plane_euler_MC'], 'Euler MC')
    df_inf['p_mc_cen'] = (p_mc + 180) % 360 - 180
else:
    df_inf['p_mc_cen'] = df_inf['p_rec_cen']

phi_bin_edges = np.linspace(-180, 180, 13)
phi_centers = (phi_bin_edges[:-1] + phi_bin_edges[1:]) / 2

# 4. Las 5 Filas Definitivas
configs = [
    ('nMuones_REC', r'$N_\mu^{\mathrm{REC}}$ (UMD)', 'navy'), 
    ('nMuones_MC', r'$N_\mu^{\mathrm{MC}}$ (UMD)', 'mediumblue'), 
    ('sdSignal_REC', r'SD Total (VEM)', 'firebrick'),
    ('sd_nMuons_MC', r'SD Muons (MC)', 'forestgreen'),
    ('sd_nEM_MC', r'SD EM (MC)', 'darkorange')
]

# 5. Preparación del Plot (5 Filas x 6 Columnas)
fig, axs = plt.subplots(5, 6, figsize=(16, 12), sharex=True, sharey='row')

for k, (col_name, label, color) in enumerate(configs):
    for j in range(6):
        ax = axs[k, j]
        df_r = df_inf[df_inf['r_bin'] == j].copy()
        
        if col_name not in df_r.columns:
            ax.text(0.5, 0.5, f"No Data", ha='center', va='center', alpha=0.5)
            continue
            
        if df_r.empty or len(df_r.dropna(subset=[col_name])) < 15:
            ax.text(0.5, 0.5, "Baja Est.", ha='center', va='center', alpha=0.5)
            continue

        # Usamos ángulo MC para las variables MC puras, REC para las reconstruidas
        angle_col = 'p_rec_cen' if 'REC' in col_name else 'p_mc_cen'
        df_r['bin_phi'] = pd.cut(df_r[angle_col], bins=phi_bin_edges)
        
        st = df_r.groupby('bin_phi', observed=True)[col_name].agg(['mean','sem']).dropna()
        
        means = st['mean']
        sems = st['sem']
        norm = means.mean()
        
        if norm == 0 or pd.isna(norm): continue
        
        # Ploteo
        ax.errorbar(phi_centers, means/norm, yerr=sems/norm, fmt='o', 
                    color=color, markersize=6, capsize=3, elinewidth=1.5)
        
        # Estética de sub-panel
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.set_xlim(-180, 180)
        ax.set_xticks([-150, -100, -50, 0, 50, 100, 150])
        
        if k == 0: 
            ax.set_title(r_labels[j], fontsize=15, fontweight='bold', pad=12)
        if k == 4: 
            ax.set_xlabel(r"Azimuth $\phi$ [deg]", fontsize=13)
            ax.tick_params(axis='x', labelsize=10)
        else:
            ax.tick_params(axis='x', labelbottom=False)
            
        if j == 0: 
            ax.set_ylabel(f"{label}\nNorm. Amp.", fontsize=14, fontweight='bold', color=color)

fig.suptitle(r"Infill Signal Decoupling: Inclined Showers ($40^\circ \leq \theta \leq 60^\circ$)", 
             fontsize=22, fontweight='bold', y=0.96)

fig.text(0.02, 0.96, r"$\mathbf{Proton\ (SIB2.3e)}$", fontsize=14, va='top', ha='left', color='dimgray')

plt.subplots_adjust(hspace=0.08, wspace=0.08)
plt.savefig('core_rec.png')
plt.show()
```

```python
# =========================================================================
# CELDA 2: PERFILES INFILL (40°-60°) - TRUE CORE EVALUATION (LORENZO'S FIX)
# =========================================================================

# 1. Filtrar Infill y Rango Zenital Inclinado
df_inf = df_guia[(df_guia['counterId'] >= 100000) & 
                 (df_guia['theta_REC'] >= 40) & 
                 (df_guia['theta_REC'] <= 60)].copy()

# 2. Configuración de Bines Radiales (6 columnas)
r_edges = [0, 200, 400, 600, 800, 1000, 1500] 
r_labels = ["0-200 m", "200-400 m", "400-600 m", "600-800 m", "800-1000 m", "1000-1500 m"]

# Creamos DOS columnas de bines radiales: una para REC y otra para MC
df_inf['r_bin_REC'] = pd.cut(df_inf['r_core'], bins=r_edges, labels=False, include_lowest=True)

if 'r_core_MC' in df_inf.columns:
    df_inf['r_bin_MC'] = pd.cut(df_inf['r_core_MC'], bins=r_edges, labels=False, include_lowest=True)
else:
    print("Advertencia: r_core_MC no encontrado. Usando r_core para todo.")
    df_inf['r_bin_MC'] = df_inf['r_bin_REC'] 

# 3. Pre-cálculo de ángulos con Euler Fix (Centrados en [-180, 180])
p_rec = ensure_degrees(df_inf['phi_plane_euler'], 'Euler REC')
df_inf['p_rec_cen'] = (p_rec + 180) % 360 - 180

# Usamos la nueva variable del True Core para MC
if 'phi_plane_euler_MC_true_core' in df_inf.columns:
    p_mc = ensure_degrees(df_inf['phi_plane_euler_MC_true_core'], 'Euler MC True')
    df_inf['p_mc_cen'] = (p_mc + 180) % 360 - 180
elif 'phi_plane_euler_MC' in df_inf.columns:
    p_mc = ensure_degrees(df_inf['phi_plane_euler_MC'], 'Euler MC Old')
    df_inf['p_mc_cen'] = (p_mc + 180) % 360 - 180
else:
    df_inf['p_mc_cen'] = df_inf['p_rec_cen']

phi_bin_edges = np.linspace(-180, 180, 13)
phi_centers = (phi_bin_edges[:-1] + phi_bin_edges[1:]) / 2

# 4. Las 5 Filas Definitivas
configs = [
    ('nMuones_REC', r'$N_\mu^{\mathrm{REC}}$ (UMD)', 'navy'), 
    ('nMuones_MC', r'$N_\mu^{\mathrm{MC}}$ (UMD)', 'mediumblue'), 
    ('sdSignal_REC', r'SD Total (VEM)', 'firebrick'),
    ('sd_nMuons_MC', r'SD Muons (MC)', 'forestgreen'),
    ('sd_nEM_MC', r'SD EM (MC)', 'darkorange')
]

# 5. Preparación del Plot (5 Filas x 6 Columnas)
fig, axs = plt.subplots(5, 6, figsize=(20, 14), sharex=True, sharey='row')

for k, (col_name, label, color) in enumerate(configs):
    for j in range(6):
        ax = axs[k, j]
        
        # LÓGICA CLAVE: Elegir el Bin Radial correcto (REC o MC)
        is_rec_var = 'REC' in col_name
        r_bin_col = 'r_bin_REC' if is_rec_var else 'r_bin_MC'
        
        df_r = df_inf[df_inf[r_bin_col] == j].copy()
        
        if col_name not in df_r.columns:
            ax.text(0.5, 0.5, f"No Data", ha='center', va='center', alpha=0.5)
            continue
            
        if df_r.empty or len(df_r.dropna(subset=[col_name])) < 15:
            ax.text(0.5, 0.5, "Baja Est.", ha='center', va='center', alpha=0.5)
            continue

        # Usamos ángulo MC (True Core) para las variables MC puras, REC para las reconstruidas
        angle_col = 'p_rec_cen' if is_rec_var else 'p_mc_cen'
        df_r['bin_phi'] = pd.cut(df_r[angle_col], bins=phi_bin_edges)
        
        st = df_r.groupby('bin_phi', observed=True)[col_name].agg(['mean','sem']).dropna()
        
        means = st['mean']
        sems = st['sem']
        norm = means.mean()
        
        if norm == 0 or pd.isna(norm): continue
        
        # Ploteo
        ax.errorbar(phi_centers, means/norm, yerr=sems/norm, fmt='o', 
                    color=color, markersize=6, capsize=3, elinewidth=1.5)
        
        # Estética de sub-panel
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.set_xlim(-180, 180)
        ax.set_xticks([-150, -100, -50, 0, 50, 100, 150])
        
        if k == 0: 
            ax.set_title(r_labels[j], fontsize=15, fontweight='bold', pad=12)
        if k == 4: 
            ax.set_xlabel(r"Azimuth $\phi$ [deg]", fontsize=13)
            ax.tick_params(axis='x', labelsize=10)
        else:
            ax.tick_params(axis='x', labelbottom=False)
            
        if j == 0: 
            ax.set_ylabel(f"{label}\nNorm. Amp.", fontsize=14, fontweight='bold', color=color)

fig.suptitle(r"Infill Signal Decoupling: TRUE CORE EVALUATION ($40^\circ \leq \theta \leq 60^\circ$)", 
             fontsize=22, fontweight='bold', y=0.96)

fig.text(0.02, 0.96, r"$\mathbf{Proton\ (SIB2.3e)}$ | True Geometry Applied to MC", fontsize=14, va='top', ha='left', color='dimgray')

plt.subplots_adjust(hspace=0.08, wspace=0.08)
plt.savefig('core_mc.png')
plt.show()
```

```python
# =========================================================================
# CELDA: CHEQUEO DE DESFASE CORE REC vs CORE MC (CORREGIDA)
# =========================================================================

# Filtramos lluvias inclinadas del Infill
df_cores = df_guia[(df_guia['counterId'] >= 100000) & 
                   (df_guia['theta_REC'] >= 40) & 
                   (df_guia['theta_REC'] <= 60)].copy()

# Elegimos un bin de distancia lejano donde vemos el problema (ej. 800-1000m)
df_far = df_cores[(df_cores['r_core'] > 800) & (df_cores['r_core'] < 1000)].copy()

# Calculamos la diferencia de distancia: ¿Cuánto le erró el Offline?
df_far['delta_r'] = df_far['r_core_MC'] - df_far['r_core']

# Aseguramos que el ángulo reconstruido esté en grados y centrado [-180, 180]
df_far['phi_deg'] = (np.rad2deg(df_far['phi_plane_euler']) if df_far['phi_plane_euler'].abs().max() < 7 else df_far['phi_plane_euler'])
df_far['phi_deg'] = (df_far['phi_deg'] + 180) % 360 - 180

fig, ax = plt.subplots(figsize=(10, 6))

# Ploteamos el scatter de fondo
ax.scatter(df_far['phi_deg'], df_far['delta_r'], alpha=0.3, color='purple', s=10)

# Calculamos la media por bin angular (¡EN GRADOS!)
bins = np.linspace(-180, 180, 20)
bin_centers = (bins[:-1] + bins[1:]) / 2

df_far['phi_bin'] = pd.cut(df_far['phi_deg'], bins)

# observed=False evita que se rompan las dimensiones si un bin está vacío
mean_delta_r = df_far.groupby('phi_bin', observed=False)['delta_r'].mean()

ax.plot(bin_centers, mean_delta_r.values, 'ro-', lw=2, markersize=8, label='Desfase Medio')

ax.set_xlabel(r"Azimuth $\phi$ [deg] (Reconstruido)", fontsize=14)
ax.set_ylabel(r"$r_{\mathrm{Core\ MC}} - r_{\mathrm{Core\ REC}}$ [m]", fontsize=14)
ax.set_title("Desfase entre el Core Geométrico y el Reconstruido vs. Azimut (800-1000 m)", fontsize=15)
ax.axhline(0, color='k', linestyle='--', lw=2)
ax.set_xlim(-180, 180)
ax.grid(True, alpha=0.5)
ax.legend()

plt.show()
```

```python
# =========================================================================
# CELDA 2: VALIDACIÓN GEOMÉTRICA (UMD) - DARKO VS GROUND & EULER
# =========================================================================
import numpy as np
import matplotlib.pyplot as plt

# 1. Configuración de Estilo HEP
hep_style = {
    'font.family': 'sans-serif',
    'font.sans-serif': ['Helvetica', 'Arial', 'DejaVu Sans'],
    'xtick.direction': 'in', 'ytick.direction': 'in',
    'xtick.top': True, 'ytick.right': True,
    'xtick.minor.visible': True, 'ytick.minor.visible': True,
    'axes.linewidth': 1.2,
    'legend.frameon': True, 'legend.shadow': True, 'legend.edgecolor': 'black',
    'lines.linewidth': 2, 'lines.markersize': 8
}
plt.rcParams.update(hep_style)

# Función de ayuda para diferencia angular (rango -180 a 180)
def ang_diff(a, b):
    d = a - b
    return (d + 180) % 360 - 180

# --- PROCESAMIENTO DE DATOS ---
df_infill = df_guia[df_guia['counterId'] >= 100000].copy()

if len(df_infill) > 0:
    # --- 1. Extraemos las variables y convertimos a grados ---
    vals_ground = ensure_degrees(df_infill['phi_plane_ground'], 'Ground')
    vals_euler = ensure_degrees(df_infill['phi_plane_euler'], 'EulerREC')
    vals_darko_rec = ensure_degrees(df_infill['phi_plane_darko_rec'], 'DarkoRec')
    vals_theta_rec = ensure_degrees(df_infill['theta_REC'], 'ThetaRec')

    # --- 2. Cálculos de Diferencias ---
    # PLOT 1: Darko vs Ground
    diff_darko_ground = ang_diff(vals_darko_rec, vals_ground)
    
    # PLOT 2 y 3: Tu rotación de Euler original vs Darko
    diff_euler_darko = ang_diff(vals_euler, vals_darko_rec)

    # -------------------------------------------------------------------------
    # PLOT 1: HISTOGRAMA DE RESIDUOS (DARKO REC VS GROUND 2D)
    # -------------------------------------------------------------------------
    fig1, ax1 = plt.subplots(figsize=(9, 5))
    ax1.hist(diff_darko_ground, bins=100, range=(-5, 5), color='forestgreen', 
             alpha=0.7, edgecolor='darkgreen', label=r'$\Delta\phi$ (Darko 3D - Ground 2D)')
    ax1.axvline(0, color='black', linestyle='--', lw=1.5, alpha=0.8)
    
    ax1.set_title("UMD Coordinate Validation: Darko 3D vs. Ground 2D", fontsize=16, fontweight='bold', pad=15)
    ax1.set_xlabel(r"Angular Difference $\Delta\phi$ [deg]", fontsize=14)
    ax1.set_ylabel("Counts", fontsize=14)
    ax1.legend(fontsize=12, loc='upper right')
    ax1.grid(True, alpha=0.2)
    plt.tight_layout()
    plt.show()

    # -------------------------------------------------------------------------
    # PLOT 2: EL "BUTTERFLY" PLOT (EULER VS DARKO)
    # -------------------------------------------------------------------------
    fig2, ax2 = plt.subplots(figsize=(9, 5))
    n_sample = min(30000, len(df_infill))
    idx = np.random.choice(df_infill.index, n_sample, replace=False)
    
    ax2.scatter(vals_theta_rec.loc[idx], diff_euler_darko.loc[idx], 
                s=1.5, c='navy', alpha=0.15, rasterized=True)
    ax2.axhline(0, color='crimson', linestyle='-', lw=1, alpha=0.5)
    ax2.axhline(180, color='crimson', linestyle='--', lw=1, alpha=0.5)
    ax2.axhline(-180, color='crimson', linestyle='--', lw=1, alpha=0.5)
    
    ax2.set_title(r"Analytic Euler Rotation vs. Darko Math", fontsize=16, fontweight='bold', pad=15)
    ax2.set_xlabel(r"Zenith Angle $\theta_{\mathrm{REC}}$ [deg]", fontsize=14)
    ax2.set_ylabel(r"Residual $\Delta\phi_{\mathrm{Euler-Darko}}$ [deg]", fontsize=14)
    ax2.set_ylim(-200, 200) # Ampliamos rango por si hay desfase de 180
    ax2.grid(True, alpha=0.2)
    plt.tight_layout()
    plt.show()

    # -------------------------------------------------------------------------
    # PLOT 3: EL "SNAKE" PLOT (DEP. AZIMUTAL EULER VS DARKO)
    # -------------------------------------------------------------------------
    mask_inclined = vals_theta_rec > 40
    if mask_inclined.sum() > 100:
        idx_snake = np.random.choice(df_infill[mask_inclined].index, 
                                     min(30000, mask_inclined.sum()), replace=False)
        fig3, ax3 = plt.subplots(figsize=(12, 6))
        
        sc = ax3.scatter(vals_darko_rec.loc[idx_snake], 
                         diff_euler_darko.loc[idx_snake], 
                         c=vals_theta_rec.loc[idx_snake], 
                         cmap='viridis', s=2, alpha=0.6, rasterized=True)
        
        ax3.axhline(0, color='black', linestyle='--', lw=1, alpha=0.7)
        ax3.axhline(180, color='black', linestyle='--', lw=1, alpha=0.7)
        ax3.axhline(-180, color='black', linestyle='--', lw=1, alpha=0.7)

        ax3.set_title(r"Comparison in Inclined Showers ($\theta_{\mathrm{REC}} > 40^\circ$)", 
                      fontsize=16, fontweight='bold', pad=15)
        ax3.set_xlabel(r"Darko Azimuth $\phi_{\mathrm{Darko}}^{\mathrm{REC}}$ [deg]", fontsize=14)
        ax3.set_ylabel(r"Difference $\Delta\phi$ (Euler - Darko) [deg]", fontsize=14)
        
        cbar = plt.colorbar(sc, ax=ax3)
        cbar.set_label(r"Zenith Angle $\theta_{\mathrm{REC}}$ [deg]", fontsize=12)
        
        ax3.set_xlim(0, 360)
        ax3.set_ylim(-200, 200) # Ampliamos rango
        ax3.grid(True, alpha=0.2)
        plt.tight_layout()
        plt.show()
else:
    print("No se encontraron datos de Infill para procesar.")
```

```python
# =========================================================================
# CELDA 2: VALIDACIÓN GEOMÉTRICA (UMD) - AUDITORÍA DEL AZIMUTH SP
# =========================================================================

# Función de ayuda para diferencia angular (rango -180 a 180)
def ang_diff(a, b):
    d = a - b
    return (d + 180) % 360 - 180

# --- PROCESAMIENTO DE DATOS ---
df_infill = df_guia[df_guia['counterId'] >= 100000].copy()

if len(df_infill) > 0:
    # Aplicamos Sanity Checks
    # 1. La "Verdad" del Offline para el UMD
    vals_sp_umd = ensure_degrees(df_infill['phi_plane_sp_umd'], 'SP_UMD')
    
    # 2. La Verdad del Piso (2D)
    vals_ground = ensure_degrees(df_infill['phi_plane_ground'], 'Ground')
    
    # 3. La Verdad Analítica 3D (Usando Core MC y Ángulos MC para no tener sesgos del REC)
    vals_euler_mc = ensure_degrees(df_infill['phi_plane_euler_MC_true_core'], 'Euler_MC_True')
    
    # Variables de control
    vals_theta_rec = ensure_degrees(df_infill['theta_REC'], 'ThetaRec')

    # Cálculos de Diferencias
    # NOTA: En tu read_v15, phi_plane_sp_umd ya tiene la corrección de (- rec_phi_rad), 
    # así que no hace falta sumarle phi_REC de nuevo acá. Comparamos directo.
    diff_umdSP_ground = ang_diff(vals_sp_umd, vals_ground)
    diff_euler_umdSP = ang_diff(vals_euler_mc, vals_sp_umd)

    # -------------------------------------------------------------------------
    # PLOT 1: HISTOGRAMA DE RESIDUOS (UMD SP NATIVO VS GROUND)
    # -------------------------------------------------------------------------
    fig1, ax1 = plt.subplots(figsize=(9, 5))
    ax1.hist(diff_umdSP_ground, bins=100, range=(-5, 5), color='forestgreen', 
             alpha=0.7, edgecolor='darkgreen', label=r'$\Delta\phi$ (UMD SP - Ground)')
    ax1.axvline(0, color='black', linestyle='--', lw=1.5, alpha=0.8)
    
    ax1.set_title("UMD Coordinate Validation: Offline vs. 2D Ground", fontsize=16, fontweight='bold', pad=15)
    ax1.set_xlabel(r"Angular Difference $\Delta\phi$ [deg]", fontsize=14)
    ax1.set_ylabel("Counts", fontsize=14)
    ax1.legend(fontsize=12, loc='upper right')
    ax1.grid(True, alpha=0.2)
    plt.tight_layout()
    #plt.savefig('Validation_UMD_1_Histogram.jpg', dpi=300, bbox_inches='tight')
    plt.show()

    # -------------------------------------------------------------------------
    # PLOT 2: EL "BUTTERFLY" PLOT (EULER 3D VS UMD SP NATIVO)
    # -------------------------------------------------------------------------
    fig2, ax2 = plt.subplots(figsize=(9, 5))
    n_sample = min(30000, len(df_infill))
    idx = np.random.choice(df_infill.index, n_sample, replace=False)
    
    ax2.scatter(vals_theta_rec.loc[idx], diff_euler_umdSP.loc[idx], 
                s=1.5, c='navy', alpha=0.15, rasterized=True)
    ax2.axhline(0, color='crimson', linestyle='-', lw=1, alpha=0.5)
    
    ax2.set_title(r"Analytic 3D Rotation vs. Offline UMD Azimuth SP", fontsize=16, fontweight='bold', pad=15)
    ax2.set_xlabel(r"Zenith Angle $\theta_{\mathrm{REC}}$ [deg]", fontsize=14)
    ax2.set_ylabel(r"Residual $\Delta\phi_{\mathrm{Euler-UMD\_SP}}$ [deg]", fontsize=14)
    ax2.set_ylim(-40, 40)
    ax2.grid(True, alpha=0.2)
    plt.tight_layout()
    #plt.savefig('Validation_UMD_2_Butterfly.jpg', dpi=300, bbox_inches='tight')
    plt.show()

    # -------------------------------------------------------------------------
    # PLOT 3: EL "SNAKE" PLOT (DEP. AZIMUTAL DEL ERROR DEL OFFLINE)
    # -------------------------------------------------------------------------
    mask_inclined = vals_theta_rec > 40
    if mask_inclined.sum() > 100:
        idx_snake = np.random.choice(df_infill[mask_inclined].index, 
                                     min(30000, mask_inclined.sum()), replace=False)
        fig3, ax3 = plt.subplots(figsize=(12, 6))
        
        sc = ax3.scatter(vals_euler_mc.loc[idx_snake], 
                         diff_euler_umdSP.loc[idx_snake], 
                         c=vals_theta_rec.loc[idx_snake], 
                         cmap='viridis', s=2, alpha=0.6, rasterized=True)
        
        ax3.axhline(0, color='black', linestyle='--', lw=1, alpha=0.7)
        ax3.set_title(r"Offline UMD Error in Inclined Showers ($\theta_{\mathrm{REC}} > 40^\circ$)", 
                      fontsize=16, fontweight='bold', pad=15)
        ax3.set_xlabel(r"True Analytic Azimuth $\phi_{\mathrm{Euler\_MC}}$ [deg]", fontsize=14)
        ax3.set_ylabel(r"Error $\Delta\phi$ (Euler - UMD SP) [deg]", fontsize=14)
        
        cbar = plt.colorbar(sc, ax=ax3)
        cbar.set_label(r"Zenith Angle $\theta_{\mathrm{REC}}$ [deg]", fontsize=12)
        
        ax3.set_xlim(0, 360)
        ax3.set_ylim(-30, 30)
        ax3.grid(True, alpha=0.2)
        plt.tight_layout()
        #plt.savefig('Validation_UMD_3_Snake.jpg', dpi=300, bbox_inches='tight')
        plt.show()
else:
    print("No se encontraron datos de Infill para procesar.")
```

```python
# =========================================================================
# CELDA 3: PERFILES INFILL (40°-60°) - UMD GEOMETRY EVALUATION (DARKO)
# =========================================================================

# 1. Filtrar Infill y Rango Zenital Inclinado
df_inf = df_guia[(df_guia['counterId'] >= 100000) & 
                 (df_guia['theta_REC'] >= 40) & 
                 (df_guia['theta_REC'] <= 60)].copy()

# 2. Configuración de Bines Radiales (Usando r_umd_rec y r_umd_mc)
r_edges = [0, 200, 400, 600, 800, 1000, 1500] 
r_labels = ["0-200 m", "200-400 m", "400-600 m", "600-800 m", "800-1000 m", "1000-1500 m"]

df_inf['r_bin_REC'] = pd.cut(df_inf['r_umd_rec'], bins=r_edges, labels=False, include_lowest=True)
df_inf['r_bin_MC'] = pd.cut(df_inf['r_umd_mc'], bins=r_edges, labels=False, include_lowest=True)

# 3. Pre-cálculo de ángulos UMD Centrados en [-180, 180]
p_rec = ensure_degrees(df_inf['phi_plane_darko_rec'], 'Darko REC')
df_inf['p_rec_cen'] = (p_rec + 180) % 360 - 180

p_mc = ensure_degrees(df_inf['phi_plane_darko_mc'], 'Darko MC')
df_inf['p_mc_cen'] = (p_mc + 180) % 360 - 180

phi_bin_edges = np.linspace(-180, 180, 13)
phi_centers = (phi_bin_edges[:-1] + phi_bin_edges[1:]) / 2

# 4. Las 5 Filas
configs = [
    ('nMuones_REC', r'$N_\mu^{\mathrm{REC}}$ (UMD)', 'navy'), 
    ('nMuones_MC', r'$N_\mu^{\mathrm{MC}}$ (UMD)', 'mediumblue'), 
    ('sdSignal_REC', r'SD Total (VEM)', 'firebrick'),
    ('sd_nMuons_MC', r'SD Muons (MC)', 'forestgreen'),
    ('sd_nEM_MC', r'SD EM (MC)', 'darkorange')
]

# 5. Preparación del Plot
fig, axs = plt.subplots(5, 6, figsize=(20, 14), sharex=True, sharey='row')

for k, (col_name, label, color) in enumerate(configs):
    for j in range(6):
        ax = axs[k, j]
        
        is_rec_var = 'REC' in col_name
        r_bin_col = 'r_bin_REC' if is_rec_var else 'r_bin_MC'
        
        df_r = df_inf[df_inf[r_bin_col] == j].copy()
        
        if col_name not in df_r.columns:
            continue
            
        if df_r.empty or len(df_r.dropna(subset=[col_name])) < 15:
            ax.text(0.5, 0.5, "Baja Est.", ha='center', va='center', alpha=0.5)
            continue

        angle_col = 'p_rec_cen' if is_rec_var else 'p_mc_cen'
        df_r['bin_phi'] = pd.cut(df_r[angle_col], bins=phi_bin_edges)
        
        # observed=False para mantener las dimensiones seguras
        st = df_r.groupby('bin_phi', observed=False)[col_name].agg(['mean','sem']).dropna()
        
        means = st['mean']
        sems = st['sem']
        norm = means.mean()
        
        if norm == 0 or pd.isna(norm): continue
        
        ax.errorbar(phi_centers, means/norm, yerr=sems/norm, fmt='o', 
                    color=color, markersize=6, capsize=3, elinewidth=1.5)
        
        ax.grid(True, alpha=0.3, linestyle='--')
        ax.set_xlim(-180, 180)
        ax.set_xticks([-150, -100, -50, 0, 50, 100, 150])
        
        if k == 0: 
            ax.set_title(r_labels[j], fontsize=15, fontweight='bold', pad=12)
        if k == 4: 
            ax.set_xlabel(r"Azimuth $\phi$ [deg]", fontsize=13)
            ax.tick_params(axis='x', labelsize=10)
        else:
            ax.tick_params(axis='x', labelbottom=False)
            
        if j == 0: 
            ax.set_ylabel(f"{label}\nNorm. Amp.", fontsize=14, fontweight='bold', color=color)

fig.suptitle(r"Infill Signal Decoupling: UMD GEOMETRY (Darko's Math) | $40^\circ \leq \theta \leq 60^\circ$", 
             fontsize=22, fontweight='bold', y=0.96)

fig.text(0.02, 0.96, r"$\mathbf{Proton\ (SIB2.3e)}$", fontsize=14, va='top', ha='left', color='dimgray')

plt.subplots_adjust(hspace=0.08, wspace=0.08)
plt.show()
```

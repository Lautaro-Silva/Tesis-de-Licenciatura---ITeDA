## 📦 2. Funciones de lectura

Si `readADST_data_v19.py` está en el mismo directorio se importa automáticamente.  
De lo contrario, las funciones se definen directamente acá.

```python
try:
    from readADST_data_v19 import getModuleList, readADST_surface_data_v19
    print('Funciones importadas desde readADST_data_v19.py')
except ImportError:
    print('readADST_data_v19.py no encontrado — definiendo inline')

    def getModuleList(counter, sim=False):
        ids = range(0, 6) if sim else range(100, 116)
        return [counter.GetModule(mid) for mid in ids if counter.HasModule(mid)]

    def readADST_surface_data_v19(
        fname, only_6T5=True, min_logE=17.0,
        max_theta_deg=60.0, skip_rejected_stations=True,
    ):
        if not os.path.exists(fname):
            return pd.DataFrame()
        files = ROOT.std.vector('string')()
        files.push_back(fname)
        file1 = ROOT.RecEventFile(files)
        event = ROOT.RecEvent()
        file1.SetBuffers(event)
        data = []
        n_read = n_no_energy = n_no_t5 = n_theta_cut = n_passed = 0
        t0 = time.time()
        while file1.ReadNextEvent() == ROOT.RecEventFile.eSuccess:
            n_read += 1
            sEvent = event.GetSDEvent()
            sShower = sEvent.GetSdRecShower()
            rec_energy = sShower.GetEnergy()
            if rec_energy <= 0:
                n_no_energy += 1
                continue
            logE = np.log10(rec_energy)
            if logE < min_logE:
                n_no_energy += 1
                continue
            if only_6T5:
                if not sEvent.IsT5():
                    n_no_t5 += 1
                    continue
                n_ub  = sEvent.GetT5PriorActiveNeighbors().size()
                n_uub = sEvent.GetT5PriorActiveUUBNeighbors().size()
                if n_ub < 5 and n_uub < 5:
                    n_no_t5 += 1
                    continue
            theta = sShower.GetZenith() * 180.0 / np.pi
            if theta > max_theta_deg:
                n_theta_cut += 1
                continue
            n_passed += 1
            ev_id = event.GetEventId()
            phi   = sShower.GetAzimuth() * 180.0 / np.pi
            pc    = sShower.GetCoreSiteCS()
            mEvent = event.GetMDEvent()
            it, end = mEvent.CountersBegin(), mEvent.CountersEnd()
            while it != end:
                c    = it.__deref__()
                cid  = c.GetId()
                sdId = c.GetSdPartnerId()
                if not sEvent.HasStation(sdId):
                    it += 1
                    continue
                st = sEvent.GetStationById(sdId)
                if skip_rejected_stations and st.IsRejected():
                    it += 1
                    continue
                if c.IsRejected():
                    it += 1
                    continue
                for m in getModuleList(c, sim=False):
                    if m.IsCandidate():   s = 'candidate'
                    elif m.IsSaturated(): s = 'saturated'
                    elif m.IsRejected():  s = 'rejected'
                    elif m.IsSilent():    s = 'silent'
                    else:                 s = 'undefined'
                    data.append({
                        'event_id': ev_id,   'logE_REC': logE,
                        'theta_REC': theta,  'phi_REC': phi,
                        'core_x': pc.X(),    'core_y': pc.Y(),
                        'counterId': cid,    'sdId': sdId,
                        'moduleId': m.GetId(),
                        'nMuones_REC': m.GetNumberOfEstimatedMuons(),
                        'module_status': s,
                        'r_core': st.GetSPDistance(),
                        'r_core_err': st.GetSPDistanceError(),
                        'phi_plane_sp': st.GetAzimuthSP(),
                        'is_sd_saturated': st.IsLowGainSaturated(),
                        'sdSignal_REC': st.GetTotalSignal(),
                        'sdSignal_err': st.GetTotalSignalError(),
                        'sdMuonSignal_REC': st.GetMuonSignal(),
                    })
                it += 1
        print(f'  Leidos: {n_read} | Pasaron: {n_passed}')
        print(f'  Tirados: energia={n_no_energy}, T5={n_no_t5}, theta={n_theta_cut}')
        print(f'  Filas guardadas: {len(data)} | Tiempo: {time.time()-t0:.1f}s')
        return pd.DataFrame(data)
```

## ✂️ 5. ¿Cuánto tira cada corte?

Leemos el mismo archivo sin ningún corte y aplicamos los filtros en pandas.  
El efecto del **6T5** no se puede re-aplicar en pandas (no se guarda como columna en el DataFrame),  
pero se mide comparando contra la lectura completa con todos los cortes.

```python
print('Leyendo sin ningun corte...')
df_raw = readADST_surface_data_v19(
    TEST_FILE,
    only_6T5=False, min_logE=0.0,
    max_theta_deg=90.0, skip_rejected_stations=False,
)

n_raw  = df_raw['event_id'].nunique()
n_E    = df_raw[df_raw['logE_REC'] >= 17.0]['event_id'].nunique()
n_th   = df_raw[df_raw['theta_REC'] <= 60.0]['event_id'].nunique()
mask   = (df_raw['logE_REC'] >= 17.0) & (df_raw['theta_REC'] <= 60.0)
n_both = df_raw[mask]['event_id'].nunique()
n_all  = df['event_id'].nunique()

print()
print('  Corte                              Eventos       %')
print('  ' + '-'*46)
rows = [
    ('Sin cortes',                        n_raw,  100.0),
    ('Solo logE >= 17',                   n_E,    100*n_E/n_raw),
    ('Solo theta <= 60 deg',              n_th,   100*n_th/n_raw),
    ('logE + theta',                      n_both, 100*n_both/n_raw),
    ('logE + theta + 6T5 + rejected',     n_all,  100*n_all/n_raw),
]
for lbl, n, pct in rows:
    print(f'  {lbl:<36} {n:>6}   {pct:.1f}%')

extra = max(0, n_both - n_all)
print()
print(f'  Tirados por 6T5 + rejected (extra sobre logE+theta): {extra} eventos')
```

## 📊 6. Distribuciones de la lluvia reconstruida

```python
events = df.drop_duplicates('event_id')

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle(f'SD-750 + UMD  {YEAR}/{MONTH}/{DAY}  (con cortes)', fontsize=13, fontweight='bold')

ax = axes[0, 0]
ax.hist(events['logE_REC'], bins=30, color='steelblue', edgecolor='white', lw=0.4)
ax.set_xlabel('log$_{10}$(E / eV)')
ax.set_ylabel('Eventos')
ax.set_title('Energia reconstruida')

ax = axes[0, 1]
ax.hist(events['theta_REC'], bins=30, color='steelblue', edgecolor='white', lw=0.4)
ax.set_xlabel('theta (grados)')
ax.set_ylabel('Eventos')
ax.set_title('Angulo cenital')

ax = axes[1, 0]
ax.hist(df['r_core'], bins=50, color='steelblue', edgecolor='white', lw=0.4)
ax.set_xlabel('r (m)')
ax.set_ylabel('Modulos UMD')
ax.set_title('Distancia al core')

ax = axes[1, 1]
ax.hist(df['phi_plane_sp'], bins=36, color='steelblue', edgecolor='white', lw=0.4)
ax.set_xlabel('phi en el plano de lluvia (rad)')
ax.set_ylabel('Modulos UMD')
ax.set_title('Azimuth relativo (deberia ser plano si OK)')

plt.tight_layout()
plt.show()
```

## 🎯 7. Estado de los módulos UMD

Los `moduleId` deben estar en 100-115 (datos reales). Si aparecen IDs 0-5 → estás leyendo simulaciones.  
El histograma de muones se hace solo sobre módulos **candidatos** (los que la reconstrucción usa).

```python
status_counts = df['module_status'].value_counts()

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('UMD — estado y signal', fontsize=13, fontweight='bold')

ax = axes[0]
color_map = {'candidate': 'steelblue', 'saturated': 'orange',
             'rejected': 'tomato', 'silent': 'gray', 'undefined': 'lightgray'}
bar_colors = [color_map.get(s, 'lightgray') for s in status_counts.index]
status_counts.plot(kind='bar', ax=ax, color=bar_colors, edgecolor='white')
ax.set_title('Estado de los modulos')
ax.set_ylabel('Modulos')
ax.tick_params(axis='x', rotation=30)

ax = axes[1]
ax.hist(df['moduleId'], bins=range(99, 118), color='steelblue',
        edgecolor='white', lw=0.4, rwidth=0.8)
ax.set_title('IDs de modulos activos\n(esperado: 100-115)')
ax.set_xlabel('moduleId')
ax.set_ylabel('Modulos')
ax.xaxis.set_major_locator(ticker.MultipleLocator(3))

ax = axes[2]
nmu = df[df['module_status'] == 'candidate']['nMuones_REC']
ax.hist(nmu, bins=50, color='steelblue', edgecolor='white', lw=0.4)
ax.set_yscale('log')
ax.set_title('Muones por modulo\n(solo candidatos)')
ax.set_xlabel('nMuones_REC')
ax.set_ylabel('Modulos (escala log)')

plt.tight_layout()
plt.show()

print('Estado de los modulos:')
total = status_counts.sum()
for s, cnt in status_counts.items():
    print(f'  {s:<12}  {cnt:>6}  ({100*cnt/total:.1f}%)')
```

## ✅ 8. Sanity checks automáticos

Resultado esperado: todos `[OK  ]`. Si alguno dice `[WARN]`, revisalo antes de procesar más datos.

```python
OK   = 'OK  '
WARN = 'WARN'

print('=== Sanity Checks ===')
print()

# 1. NaN
nan_cols = df.isnull().sum()
nan_cols = nan_cols[nan_cols > 0]
if len(nan_cols) == 0:
    print(f'[{OK}] Sin NaN en ninguna columna')
else:
    print(f'[{WARN}] Columnas con NaN:')
    print(nan_cols.to_string())

# 2. IDs de modulos: 100-115 = datos reales, 0-5 = simulaciones
mod_ids = sorted(df['moduleId'].unique())
bad_ids = [m for m in mod_ids if m < 100 or m > 115]
if not bad_ids:
    print(f'[{OK}] IDs de modulos en rango 100-115 — datos reales confirmado')
else:
    print(f'[{WARN}] IDs fuera de rango 100-115 (mezcla con sim?): {bad_ids}')

# 3. r_core positivo
n_bad_r = (df['r_core'] <= 0).sum()
tag = OK if n_bad_r == 0 else WARN
print(f'[{tag}] r_core <= 0: {n_bad_r} filas (esperado: 0)')

# 4. logE en rango razonable para SD-750
n_bad_E = ((df['logE_REC'] < 17.0) | (df['logE_REC'] > 20.5)).sum()
tag = OK if n_bad_E == 0 else WARN
print(f'[{tag}] logE fuera de [17, 20.5]: {n_bad_E} filas (esperado: 0)')

# 5. theta en rango
n_bad_t = ((df['theta_REC'] < 0) | (df['theta_REC'] > 60.0)).sum()
tag = OK if n_bad_t == 0 else WARN
print(f'[{tag}] theta fuera de [0, 60 deg]: {n_bad_t} filas (esperado: 0)')

# 6. SD signal no negativa
n_neg = (df['sdSignal_REC'] < 0).sum()
tag = OK if n_neg == 0 else WARN
print(f'[{tag}] sdSignal_REC < 0: {n_neg} filas (esperado: 0)')

# Resumen de cardinalidad
n_ev  = df['event_id'].nunique()
n_ctr = df['counterId'].nunique()
n_row = len(df)
print()
print(f'[{OK}] Eventos unicos:            {n_ev}')
print(f'[{OK}] Counters UMD unicos:       {n_ctr}')
print(f'[{OK}] Filas totales (modulos):   {n_row}')
print(f'[{OK}] Modulos por evento (avg):  {n_row/n_ev:.1f}')
print(f'[{OK}] Modulos por counter (avg): {n_row/n_ctr:.1f}')
```

```python
# --- Celda 1: Importaciones ---
import pandas as pd
import os

# Configuración para que pandas muestre más columnas
pd.set_option('display.max_columns', 50)
```

```python
# --- Celda 2: Configuración de Archivos ---


base_path = "/home/lsilva/Github/carmina_csvs/" 

# 2. El set específico que queremos probar
# (Ej: EPOS-LHC (R), Protón, 17.5-18.0)
file_prefix = "EPOSLHC_R_Proton_17.5_18.0"

# 3. Construir las rutas completas
counters_file = os.path.join(base_path, f"{file_prefix}_counters.csv")
modules_file = os.path.join(base_path, f"{file_prefix}_modules.csv")

print(f"Cargando el set: {file_prefix}")
print(f"Archivo de Counters: {counters_file}")
print(f"Archivo de Modules: {modules_file}")
```

```python
# --- Celda 3: Cargar los CSVs en DataFrames ---

print(f"Leyendo {os.path.basename(counters_file)}...")
try:
    df_counters = pd.read_csv(counters_file)
    print("  ¡'df_counters' cargado con éxito!")
    print(f"  Filas (counters): {len(df_counters)}")
except FileNotFoundError:
    print(f"  ¡ERROR! No se encontró el archivo. Verifica la ruta en la Celda 2.")
except Exception as e:
    print(f"  ¡ERROR inesperado!: {e}")

print(f"\nLeyendo {os.path.basename(modules_file)}...")
try:
    df_modules = pd.read_csv(modules_file)
    print("  ¡'df_modules' cargado con éxito!")
    print(f"  Filas (modules): {len(df_modules)}")
except FileNotFoundError:
    print(f"  ¡ERROR! No se encontró el archivo. Verifica la ruta en la Celda 2.")
except Exception as e:
    print(f"  ¡ERROR inesperado!: {e}")
```

```python
# --- Celda 4: Inspeccionar los DataFrames ---

# (Asegúrate de haber corrido la Celda 3 sin errores antes de correr esta)

print("--- COUNTERS ---")
if 'df_counters' in locals():
    print(df_counters.head())
    print("\n--- Info (columnas y tipos) ---")
    df_counters.info()
else:
    print("El DataFrame 'df_counters' no fue cargado. Revisa la Celda 3.")

print("\n" + "="*40 + "\n") # Separador

print("--- MODULES ---")
if 'df_modules' in locals():
    print(df_modules.head())
    print("\n--- Info (columnas y tipos) ---")
    df_modules.info()
else:
    print("El DataFrame 'df_modules' no fue cargado. Revisa la Celda 3.")
```

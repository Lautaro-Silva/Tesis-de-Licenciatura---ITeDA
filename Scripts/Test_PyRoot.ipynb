```python
import ROOT
import os
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Verifica que AUGEROFFLINEROOT está definido
print("AUGEROFFLINEROOT =", os.environ.get("AUGEROFFLINEROOT"))

# Carga la librería de Offline para acceder a RecEvent
ROOT.gSystem.Load(os.path.join(os.environ["AUGEROFFLINEROOT"], "lib/libRecEventKG.so"))
```

```python
fname = "ADST_SUPERAMIGA_230cm_Proton_e1850_t15_v2.root"

# Abrimos el TFile
f = ROOT.TFile.Open(fname)
print(f)

# Ver qué TTrees y objetos tiene
f.ls()
```

```python
# Creamos un objeto RecEvent (Offline)
event = ROOT.RecEvent()

# Abrimos el TTree principal
tree = f.Get("recData")  # también podés usar eventInfo si te interesa info de eventos
print(tree)

# SetBuffers permite leer los eventos al objeto event
tree.SetBranchAddress("event", event)  # normalmente la rama principal se llama 'event', podemos ajustar si es otra

# Leemos el primer evento
tree.GetEntry(0)

# Miramos qué hay dentro
sEvent = event.GetSDEvent()
print(sEvent)
```

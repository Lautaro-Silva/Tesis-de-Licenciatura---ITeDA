```python
f = ROOT.TFile("ADST_SUPERAMIGA_230cm_Proton_e1850_t15_v2.root")
f.ls()
```

```python
# --- 3. Acceder al TTree de reconstrucción (REC) ---
tree = f.Get("recData")
print(f"\nNúmero de eventos en recData: {tree.GetEntries()}")
tree.Print()  # imprime ramas disponibles

# --- 4. Iterar sobre algunos eventos y extraer MC y REC ---
max_events = 5  # solo los primeros 5 eventos
mc_energies = []
rec_energies = []
zenith_angles = []

for i, event in enumerate(tree):
    if i >= max_events:
        break
    
    # --- MC parameters (si existen) ---
    if hasattr(event, "GenShower"):
        mc = event.GenShower
        mc_energy = mc.GetEnergy()
        mc_zenith = mc.GetZenith() * 180.0 / np.pi  # a grados
        mc_energies.append(mc_energy)
        print(f"Evento {i} - MC Energy: {mc_energy:.2e} eV, Zenith: {mc_zenith:.2f} deg")
    
    # --- REC parameters ---
    if hasattr(event, "SdRecShower"):
        rec = event.SdRecShower
        rec_energy = rec.GetEnergy()
        rec_zenith = rec.GetZenith() * 180.0 / np.pi
        rec_energies.append(rec_energy)
        zenith_angles.append(rec_zenith)
        print(f"Evento {i} - REC Energy: {rec_energy:.2e} eV, Zenith: {rec_zenith:.2f} deg")

# --- 5. Graficar distribuciones simples ---
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(mc_energies, bins=10, alpha=0.7, label="MC")
plt.hist(rec_energies, bins=10, alpha=0.7, label="REC")
plt.xlabel("Energy [eV]")
plt.ylabel("Counts")
plt.legend()

plt.subplot(1,2,2)
plt.hist(zenith_angles, bins=10, color="orange")
plt.xlabel("Zenith angle [deg]")
plt.ylabel("Counts")

plt.tight_layout()
plt.show()
```
